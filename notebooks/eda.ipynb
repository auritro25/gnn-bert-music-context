# EDA — GNN–BERT Music Context
Run after creating `data/splits/musiccaps.csv`.

```python
import ast, pandas as pd, matplotlib.pyplot as plt
df = pd.read_csv('../data/splits/musiccaps.csv')
display(df.head())
print(df['split'].value_counts())
```

```python
from collections import Counter
def parse(x):
    try: return ast.literal_eval(x)
    except: return []
train = df[df['split']=='train']
c = Counter()
for x in train['tags']: c.update(parse(x))
top = pd.DataFrame(c.most_common(20), columns=['tag','count'])
display(top)
top.sort_values('count').plot.barh(x='tag', y='count', legend=False, figsize=(8,7));
plt.title('Top training tags'); plt.tight_layout();
```

```python
import torch, networkx as nx
from torch_geometric.utils import to_networkx
from pathlib import Path
sample_id = train.iloc[0]['track_id']
g = torch.load(Path('../data/processed/graphs')/f'{sample_id}.pt', weights_only=False)
G = to_networkx(g, to_undirected=True)
nx.draw(G, with_labels=True)
plt.title(f'Segment graph: {sample_id}')
```
